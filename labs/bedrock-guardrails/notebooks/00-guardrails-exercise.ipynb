## Create an inference profile



Use the Bedrock APIs or the AWS CLI to create an inference profile for a model that does not support direct on-demand access.



The next code cell prints a ready-to-run AWS CLI command. Run it manually only when the AWS account and region are confirmed.


```python
cmd = [
    'aws','bedrock','create-inference-profile',
    '--region','us-east-1',
    '--inference-profile-name','exercise3-inference-profile',
    '--model-source','{"copyFrom": "arn:aws:bedrock:us-east-1::foundation-model/amazon.nova-micro-v1:0"}'
]
print('Run manually if needed:')
print(' '.join(cmd))
```
