## Invoke the model

Use boto3 to call Amazon Bedrock and invoke a foundation model with a prompt.

```python
import json
region = 'us-east-1'
model_id = 'amazon.titan-text-express-v1'
prompt = 'Write a short welcome message for an Amazon Bedrock demo.'
bedrock_runtime = boto3.client('bedrock-runtime', region_name=region)
body = {
    'inputText': prompt,
    'textGenerationConfig': {
        'maxTokenCount': 200,
        'temperature': 0.7,
        'topP': 0.9
    }
}
print('Prepared request for model:', model_id)
print(json.dumps(body, indent=2))
```
